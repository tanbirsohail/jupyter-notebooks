## Basic Text Generation

Now that the API key is configured, let's try generating some text.

First, we'll initialize the generative model. We'll use `gemini-pro` for this example.

````python
# Cell 4: Initialize the model
try:
    model = genai.GenerativeModel('gemini-pro')
    print("Model 'gemini-pro' initialized successfully.")
except Exception as e:
    print(f"Error initializing model: {e}. Ensure your API key is configured correctly and you have access to the model.")
    model = None # Set model to None if initialization fails
````

Now, let's send a simple prompt to the model and get a response.

````python
# Cell 5: Send a prompt and get a response
if model:
    try:
        prompt = "What are the top 3 benefits of learning Python?"
        response = model.generate_content(prompt)
        
        print("Prompt:")
        print(prompt)
        print("\nResponse:")
        print(response.text)
    except Exception as e:
        print(f"Error generating content: {e}")
else:
    print("Model not initialized. Please check previous steps for errors.")
````

## Chat Functionality

Gemini models also support conversational interactions. You can start a chat session and send messages back and forth.

````python
# Cell 6: Start a chat session and send messages
if model:
    try:
        chat = model.start_chat(history=[])
        print("Chat session started.")

        prompt1 = "Hello! Can you tell me a fun fact about space?"
        print(f"User: {prompt1}")
        response1 = chat.send_message(prompt1)
        print(f"Gemini: {response1.text}")

        prompt2 = "That's interesting! What's another one related to planets?"
        print(f"\nUser: {prompt2}")
        response2 = chat.send_message(prompt2)
        print(f"Gemini: {response2.text}")
        
        # You can also access the chat history
        # print("\nChat History:")
        # for message in chat.history:
        #     print(f"{message.role}: {message.parts[0].text}")
            
    except Exception as e:
        print(f"Error during chat session: {e}")
else:
    print("Model not initialized. Please check previous steps for errors.")
````

## Streaming Responses

The Gemini API can also stream responses, allowing you to receive parts of the generated text as they become available. This is useful for long-running generations or for creating more interactive experiences.

````python
# Cell 7: Demonstrate streaming responses
if model:
    try:
        prompt_stream = "Tell me a short story about a brave knight and a friendly dragon."
        print(f"User: {prompt_stream}")
        print("\nGemini (streaming):")
        response_stream = model.generate_content(prompt_stream, stream=True)
        
        for chunk in response_stream:
            print(chunk.text, end="") # Print each chunk as it arrives
        print() # For a final newline
            
    except Exception as e:
        print(f"Error during streaming generation: {e}")
else:
    print("Model not initialized. Please check previous steps for errors.")
````
